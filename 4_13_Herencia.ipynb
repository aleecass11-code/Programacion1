# Introducción a la Herencia

En nuestro viaje por el mundo de la Programación Orientada a Objetos (POO), hemos creado clases que representan entidades en una biblioteca, como `Autor` y `Libro`. Pero, ¿qué pasa si queremos crear nuevas clases que compartan características con las ya existentes, pero con algunas diferencias o adicionales? ¡Aquí es donde entra en juego la herencia!

### Pregunta problema: 
¿Cómo podemos extender nuestras clases existentes para representar diferentes tipos de autores o libros, sin tener que reescribir todo el código?



```python
# Definimos la clase base Autor
class Autor:
    def __init__(self, nombre):
        self.nombre = nombre
```

## Conceptos clave de la herencia

La **herencia** permite que una clase (llamada subclase o clase hija) herede atributos y métodos de otra clase (llamada superclase o clase base). La herencia facilita la reutilización y extensión del código.

- **Herencia simple:** Una clase deriva de una única clase base.
- **Herencia múltiple:** Una clase puede derivar de varias clases base.
- **super():** Función que permite llamar a métodos de la clase base desde la clase derivada.

```python
# Creamos una subclase Escritor que hereda de Autor

class Autor:
    def __init__(self, nombre):
        self.nombre = nombre

class Escritor(Autor):
    def __init__(self, nombre, genero):
        super().__init__(nombre)
        self.genero = genero
        
# Instanciamos un objeto de la clase Escritor
escritor = Escritor("Mario Benedetti", "Realismo Social")
print(escritor.nombre, escritor.genero)
```

## Ejemplos prácticos de Herencia

Vamos a extender nuestra clase `Autor` para crear una nueva clase `Escritor` que tenga un atributo adicional: el género literario principal en el que escribe.


```python
# Instanciamos un objeto de la clase Escritor
escritor = Escritor("Mario Benedetti", "Realismo Social")
print(escritor.nombre, escritor.genero)
```

## Herencia Múltiple

En Python, una clase puede heredar de varias clases base, lo que se conoce como herencia múltiple. Aunque puede ser útil, también puede llevar a complicaciones si varias clases base tienen atributos o métodos con el mismo nombre.


```python
# Definimos una segunda clase base
class Academico:
    def __init__(self, universidad):
        self.universidad = universidad

# Creamos una clase que hereda de Escritor y Academico
class EscritorAcademico(Escritor, Academico):
    def __init__(self, nombre, genero, universidad):
        Escritor.__init__(self, nombre, genero)
        Academico.__init__(self, universidad)
```

```python
class Libro:
    def __init__(self, titulo="", isbn="", autor=None):
        self.__titulo = titulo
        self.__isbn = isbn
        self.__autor = autor

    # Métodos getter
    def get_titulo(self):
        return self.__titulo

    def get_isbn(self):
        return self.__isbn
    
    def get_autor(self):
        return self.__autor

    # Métodos setter
    def set_titulo(self, titulo):
        self.__titulo = titulo

    def set_isbn(self, isbn):
        self.__isbn = isbn

    def set_autor(self, autor):
        self.__autor = autor

#Desafío 59
class LibroDigital(Libro):
    def __init__(self, titulo="", isbn="", autor=None, formato="", tamano_archivo=0):
        super().__init__(titulo, isbn, autor)
        self.formato = formato
        self.tamano_archivo = tamano_archivo
        
    def mostrar_informacion(self):
        print(f"Título: {self.get_titulo()}")
        print(f"ISBN: {self.get_isbn()}")
        print(f"Autor: {self.get_autor().nombre if self.get_autor() else 'Desconocido'}")
        print(f"Formato: {self.formato}")
        print(f"Tamaño del archivo: {self.tamano_archivo} MB")
        
class EBook(LibroDigital):
    def __init__(self, titulo="", isbn="", autor=None, formato="", tamano_archivo=0, enlace_descarga=""):
        super().__init__(titulo, isbn, autor, formato, tamano_archivo)
        self.enlace_descarga = enlace_descarga
        
    def mostrar_informacion(self):
        super().mostrar_informacion()
        print(f"Enlace de descarga: {self.enlace_descarga}")
    


class Autor:
    def __init__(self, nombre="", nacionalidad="", libros=None):
        self.nombre = nombre
        self.nacionalidad = nacionalidad
        self.libros = libros if libros is not None else []  # Lista de libros del autor

    @property
    def nombre(self):
        return self.__nombre

    @nombre.setter
    def nombre(self, nombre):
        if not isinstance(nombre, str) or not nombre.strip():
            raise ValueError
        self.__nombre = nombre 

    @property
    def nacionalidad(self):
        return self.__nacionalidad

    @nacionalidad.setter
    def nacionalidad(self, nacionalidad):
        if not isinstance(nacionalidad, str) or not nacionalidad.strip():
            raise ValueError
        self.__nacionalidad = nacionalidad
        
        
    def agregar_libro(self, titulo, isbn):
        nuevo_libro = Libro(titulo, isbn, autor=self)
        self.libros.append(nuevo_libro)
        return nuevo_libro
    
    def eliminar_libro(self, libro):
        if libro in self.libros:
            self.libros.remove(libro)
            libro.set_autor(None)
            print(f"El libro '{libro.get_titulo()}' ha sido removido del autor.")
        else:
            print(f"El libro '{libro.get_titulo()}' no se encuentra en la lista de este autor.")
            
    def lista_titulos(self):
        """Devuelve una lista con los títulos de los libros del autor."""
        return [libro.get_titulo() for libro in self.libros]
    
    def agregar_libro_digital(self, titulo, isbn, formato, tamano_archivo):
        nuevo_libro = LibroDigital(titulo, isbn, autor=self, formato=formato, tamano_archivo=tamano_archivo)
        self.libros.append(nuevo_libro)
        return nuevo_libro
    
    def agregar_ebook(self, titulo, isbn, formato, tamano_archivo, enlace_descarga):
        nuevo_ebook = EBook(titulo, isbn, autor=self, formato=formato, tamano_archivo=tamano_archivo, enlace_descarga=enlace_descarga)
        self.libros.append(nuevo_ebook)
        return nuevo_ebook
    
        
autor1 = Autor("Pablo Neruda", "Chileno")

autor1.agregar_libro_digital("Veinte poemas de amor y una canción desesperada", "978-84-376-0494-7", "PDF", 8)
autor1.agregar_libro_digital("El habitante y su sombra", "978-84-376-0495-4", "EPUB", 10)
autor1.agregar_ebook("Cien sonetos de amor", "978-84-376-0493-0", "MOBI", 12, "https://ejemplo.com/cien-sonetos-de-amor")
autor1.agregar_ebook("Residencia en la tierra", "978-84-376-0495-4", "EPUB", 10, "https://ejemplo.com/residencia-en-la-tierra")


for libro in autor1.libros:
    if type (libro) is LibroDigital:
        libro.mostrar_informacion()
        print("-" * 30)
        
for ebook in autor1.libros:
    if type (ebook) is EBook:
        ebook.mostrar_informacion()
        print("-" * 30)
```

```python
#Desafío 60
class Escritor:
    def __init__(self, nombre, nacionalidad, genero, **kwargs):
        super().__init__(**kwargs)  #Delega los parámetros restantes a Academico
        self.nombre = nombre
        self.nacionalidad = nacionalidad
        self.genero = genero


class Academico:
    def __init__(self, especialidad, institucion, **kwargs):
        super().__init__(**kwargs)
        self.especialidad = especialidad
        self.institucion = institucion


class EscritorAcademico(Escritor, Academico):
    def __init__(self, nombre, nacionalidad, genero, especialidad, institucion):
        super().__init__(nombre=nombre, nacionalidad=nacionalidad, genero=genero, especialidad=especialidad, institucion=institucion,)

    def publicar_articulo(self, titulo, revista):
        print(f"{self.nombre} ha publicado el artículo '{titulo}' en {revista}.")


articulo1 = EscritorAcademico ("Dr. Juan Pérez", "Uruguayo", "Ensayo", "Literatura Comparada", "UDELAR")
articulo1.publicar_articulo("Análisis de Textos", "Revista UDELAR")
```

```python
#Desafío 61
class Empleado:
    def __init__(self, nombre, contraseña):
        self.nombre = nombre
        self._contraseña = contraseña
        
    def acceso_sistema(self, contraseña):
        if self._contraseña == contraseña:
            print(f"Iniciando sesión {self.nombre}.")
        else:
            print("Contraseña incorrecta. Acceso denegado.")          
        
class Administrador:
    def __init__(self, permisos):
        self.permisos = permisos
        
class Mantenimiento:
    def __init__(self, herramientas):
        self.herramientas = herramientas

class Gerente(Empleado):
    def __init__(self, nombre, _contraseña, area, permisos):
        super().__init__(nombre, _contraseña)
        self.area = area
        self.rol_administrativo = Administrador(permisos)
        
class Tecnico(Empleado):
    def __init__(self, nombre, _contraseña, especialidad, herramientas):
        super().__init__(nombre, _contraseña)
        self.especialidad = especialidad
        self.rol_mantenimiento = Mantenimiento(herramientas)

class UsuarioSinAcceso(Exception):
    pass
        
class Voluntario():
    def __init__(self, nombre, horas_disponibles):
        self.nombre = nombre
        self.horas_disponibles = horas_disponibles
        
    def __getattr__(self, atributo):
        if atributo == "acceso_sistema":
            raise UsuarioSinAcceso(f"{self.nombre} no tiene acceso al sistema. Acceso denegado.")

  
#Prueba
empleado1 = Empleado("Ana", "1234")
#Contraseña correcta
empleado1.acceso_sistema("1234")
print()
#Contraseña incorrecta
empleado1.acceso_sistema("4321")
print()

voluntario1 = Voluntario("Juan", 10)
try:
    voluntario1.acceso_sistema("Juan", "1234")
except UsuarioSinAcceso:
    print(f"{voluntario1.nombre} no tiene acceso al sistema. Acceso denegado.")
    
gerente1 = Gerente("Carlos", "5678", "Administración", ["Gestión de personal", "Control financiero"])
gerente1.acceso_sistema("5678")

tecnico1 = Tecnico("Luis", "abcd", "Sistemas", ["Destornillador", "Multímetro"])
tecnico1.acceso_sistema("abcd")

print()
print(f'Permisos de {gerente1.nombre} {gerente1.rol_administrativo.permisos}')
print()
print(f'Herramientas de {tecnico1.nombre} {tecnico1.rol_mantenimiento.herramientas}')



    
```
