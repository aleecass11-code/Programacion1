## Introducción al Encapsulamiento y acceso a miembros

El encapsulamiento es uno de los cuatro pilares fundamentales de la Programación Orientada a Objetos (POO). Se refiere a la restricción del acceso a ciertos componentes de un objeto, asegurando que solo se pueda acceder a ciertos atributos y métodos de un objeto desde fuera de la clase de la manera deseada.

### Pregunta problema: 
Imagina que estás diseñando una clase Autor para una biblioteca. ¿Cómo puedes garantizar que ciertos atributos, como el nombre, la nacionalidad o la fecha de nacimiento, no sean modificados accidentalmente o accedidos de manera inapropiada?

## Conceptos clave

_Encapsulamiento:_ Es la técnica de hacer que los campos de una clase sean privados o protegidos y proporcionar acceso a través de métodos públicos.

Diferencias entre "_" y "__" en Python:

_: Se utiliza para indicar que un atributo o método es protegido. Aunque técnicamente aún es accesible desde fuera de la clase, por convención se entiende que no debe ser accedido directamente. Sin embargo, esta protección es solo una convención y no impide realmente el acceso.

__: Se utiliza para indicar que un atributo o método es privado. Esto modifica el nombre del atributo/método de manera que es más difícil de acceder desde fuera de la clase, proporcionando una capa adicional de seguridad.

## Ejemplos prácticos

### Clase Autor con atributos públicos:

```python
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.nombre = nombre
        self.nacionalidad = nacionalidad

    def mostrar_autor(self):
        print(f"Nombre: {self.nombre}")
        print(f"Nacionalidad: {self.nacionalidad}")

autor1 = Autor("Roy Berocay", "Uruguayo")
autor1.mostrar_autor()
```

Este diseño, aunque funcional, no protege los atributos nombre y nacionalidad de modificaciones no deseadas.

### Clase Autor con atributos protegidos:

```python
class Autor:
    def __init__(self, nombre, nacionalidad):
        self._nombre = nombre
        self._nacionalidad = nacionalidad
        
    def mostrar_autor(self):
        print(f"Nombre: {self._nombre}")
        print(f"Nacionalidad: {self._nacionalidad}")
        
autor1 = Autor("Roy Berocay", "Uruguayo")
autor1.mostrar_autor()
```

Aquí, hemos marcado los atributos como protegidos usando un solo guion bajo. Esto indica que no deben ser accedidos directamente, pero aún es posible hacerlo.

### Clase Autor con atributos privados y métodos getter y setter:

```python
class Autor:
    def __init__(self, nombre, nacionalidad):
        self.__nombre = nombre
        self.__nacionalidad = nacionalidad

    def get_nombre(self):
        return self.__nombre

    def set_nombre(self, nombre):
        self.__nombre = nombre

    def get_nacionalidad(self):
        return self.__nacionalidad

    def set_nacionalidad(self, nacionalidad):
        self.__nacionalidad = nacionalidad


autor1 = Autor("Roy Berocay", "Uruguayo")
print(autor1.get_nombre())
print(autor1.get_nacionalidad())
```

Con los atributos marcados como privados, ahora es mucho más difícil acceder a ellos desde fuera de la clase. Los métodos getter y setter nos permiten controlar cómo se accede y modifica la información.

```python
#52: Crea una clase Libro que tenga atributos privados para el título, autor y ISBN. Proporciona métodos getter y setter para cada atributo.

class Libro:
    def __init__(self, titulo="", isbn="", autor=None):
        self.__titulo = titulo
        self.__isbn = isbn
        self.__autor = autor

    # Métodos getter
    def get_titulo(self):
        return self.__titulo

    def get_isbn(self):
        return self.__isbn
    
    def get_autor(self):
        return self.__autor

    # Métodos setter
    def set_titulo(self, titulo):
        self.__titulo = titulo

    def set_isbn(self, isbn):
        self.__isbn = isbn

    def set_autor(self, autor):
        self.__autor = autor


class Autor:
    def __init__(self, nombre="", nacionalidad="", libros=None):
        self.__nombre = nombre
        self.__nacionalidad = nacionalidad
        self.libros = libros if libros is not None else []  # Lista de libros del autor

    @property
    def nombre(self):
        return self.__nombre

    @nombre.setter
    def nombre(self, nombre):
        if not nombre:
            return
        self.__nombre = nombre 

    @property
    def nacionalidad(self):
        return self.__nacionalidad

    @nacionalidad.setter
    def nacionalidad(self, nacionalidad):
        if not nacionalidad:
            return
        self.__nacionalidad = nacionalidad
        
        
            
    def agregar_libro(self, titulo, isbn):
        nuevo_libro = Libro(titulo, isbn, autor=self)
        self.libros.append(nuevo_libro)
        return nuevo_libro
    
    def eliminar_libro(self, libro):
        if libro in self.libros:
            self.libros.remove(libro)
            libro.set_autor(None)
            print(f"El libro '{libro.get_titulo()}' ha sido removido del autor.")
        else:
            print(f"El libro '{libro.get_titulo()}' no se encuentra en la lista de este autor.")
            
    def lista_titulos(self):
        """Devuelve una lista con los títulos de los libros del autor."""
        if not self.libros:
            print("Sin títulos registrados")
        else:
            return [libro.get_titulo() for libro in self.libros]

def titulos_del_autor(autor):
    return autor.lista_titulos()


autor1 = Autor("Mario Benedetti", "Uruguayo")

autor1.agregar_libro("La borra del café", "1234567890")
autor1.agregar_libro("La tregua", "0987654321")
autor1.agregar_libro("Gracias por el fuego", "1122334455")
autor1.lista_titulos()

#Prueba libro
libro_prueba = Libro("Ficciones", "978-8437604947")
print("get_titulo():", libro_prueba.get_titulo())   # Ficciones
print("get_isbn():", libro_prueba.get_isbn())       # 978-8437604947
print("get_autor():", libro_prueba.get_autor())     # None

print()
libro_prueba.set_titulo("El Aleph")
libro_prueba.set_isbn("978-0307950925")
print("Nuevo título:", libro_prueba.get_titulo())  #El Aleph
print("Nuevo ISBN:", libro_prueba.get_isbn())      #978-0307950925
print()


#Prueba autor
autor_prueba = Autor("Jorge Luis Borges", "Argentina")

print("Nombre:", autor_prueba.nombre)             #Jorge Luis Borges
print("Nacionalidad:", autor_prueba.nacionalidad) #Argentina

autor_prueba.nombre = "J. L. Borges"
autor_prueba.nacionalidad = "Argentino"
print("Nombre modificado:", autor_prueba.nombre)        #J. L. Borges
print("Nacionalidad modificada:", autor_prueba.nacionalidad)  #Argentino


#Prueba autor-libros
autor_prueba.lista_titulos()  #Sin títulos registrados

libro1 = autor_prueba.agregar_libro("Ficciones", "1111")
libro2 = autor_prueba.agregar_libro("El Aleph", "2222")

print("\nAutor asignado al libro <", libro1.get_titulo(), ">: ", libro1.get_autor().nombre) #J. L. Borges

print("\nlista_titulos():", autor_prueba.lista_titulos())
print("titulos_del_autor():", titulos_del_autor(autor_prueba))

autor_prueba.eliminar_libro(libro1)
print("Autor de libro1 tras eliminar:", libro1.get_autor()) #None
print("Títulos restantes:", autor_prueba.lista_titulos())

# Eliminar un libro inexistente en el autor
print("\nEliminando libro ajeno:")
libro_ajeno = Libro("Rayuela", "3333")
autor_prueba.eliminar_libro(libro_ajeno)

```

```python
#54: Implementa la clase Autor con métodos getter y setter utilizando decoradores @property para manejar los atributos 
# privados nombre y nacionalidad.

autor1 = Autor("Gabriel García Márquez", "Colombiana")

libro1 = autor1.agregar_libro("Cien años de soledad", "978-0307474728")
libro2 = autor1.agregar_libro("El amor en los tiempos del cólera", "978-0307387264")

print(f"Autor original: {autor1.nombre}")
autor1.nombre = "G. García Márquez"         
print(f"Autor modificado: {autor1.nombre}\n")

def mostrar_catalogo(autor):
    print(f"Catálogo de {autor.nombre}:")
    titulos = autor.lista_titulos()
    if titulos:
        for titulo in titulos:
            print(f" - {titulo}")
    else:
        print(" (Sin títulos registrados)")
        
mostrar_catalogo(autor1)

autor1.eliminar_libro(libro1)

print(f"Autor de '{libro1.get_titulo()}': {libro1.get_autor()}")

autor2 = Autor("Mario Benedetti", "Uruguayo")
autor2.agregar_libro("La borra del café", "1234567890")
autor2.agregar_libro("La tregua", "0987654321")

titulos = titulos_del_autor(autor2)

print()
mostrar_catalogo(autor2)
```

```python
#Desafío 56
def autor_con_mas_libros(autors):
    autor_max_libros = autors[0]
    max_libros = len(autor_max_libros.lista_titulos())
    
    for autor in autors:
        num_libros = len(autor.lista_titulos())
        if num_libros > max_libros:
            max_libros = num_libros
            autor_max_libros = autor
    
    return autor_max_libros

autorx = Autor("Isabel Allende", "Chilena")
autorx.agregar_libro("La casa de los espíritus", "111111111")
autorx.agregar_libro("Eva Luna", "333333333")
autorx.agregar_libro("Paula", "444444444")

autorz = Autor("Jorge Luis Borges", "Argentino")
autorz.agregar_libro("Ficciones", "222222222")
autorz.agregar_libro("El Aleph", "555555555")

autor_con_mas_libros = autor_con_mas_libros([autorx, autorz])
print(f"\nEl autor con más libros es: {autor_con_mas_libros.nombre} con {len(autor_con_mas_libros.lista_titulos())} libros.")
```
