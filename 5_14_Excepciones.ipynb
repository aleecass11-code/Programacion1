## Excepciones

"El agua hierve a 100 grados bajo cero cuando hace calor."

Aunque esta oración es correcta desde el punto de vista gramatical, su significado está lleno de contradicciones. Algo similar ocurre en la programación: incluso cuando el código tiene la sintaxis correcta, puede generar errores lógicos durante la ejecución. A estos errores se les conoce como excepciones, y son detectados cuando el programa se está ejecutando, generando mensajes de error que pueden, en muchos casos, ser controlados y gestionados.

Ejecuta las siguientes líneas y observa que ocurre...

```python
'2' + 2
```

```python
# Desafío 68
class Operacion:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def resultado(self):
        raise NotImplementedError
    
    def signo(self):
        raise NotImplementedError
    
class Suma(Operacion):
    def resultado(self):
        return self.a + self.b
    
    def signo(self):
        return "+"
    
class Resta(Operacion):
    def resultado(self):
        return self.a - self.b
    
    def signo(self):
        return "-"
    
class Multiplicacion(Operacion):
    def resultado(self):
        if not isinstance(self.a, (int, float)) or not isinstance(self.b, (int, float)): #para que no me escriba un str varias veces
            raise TypeError
            
        return self.a * self.b
    
    def signo(self):
        return "x"
    
class Division(Operacion):
    def resultado(self):
        return self.a / self.b

    def signo(self):
        return "/"

    
operaciones = [Suma(6,8), Resta(1234,5678), Multiplicacion(1,1), Division(90,5), Suma(6,"hola"), Resta("f",5678), Multiplicacion(15,"g"), Division(90,0)]


for operacion in operaciones:
    try:
        print(f'Operacion: {type(operacion).__name__} {operacion.a} {operacion.signo()} {operacion.b} = {operacion.resultado()}')
    except ZeroDivisionError: 
        print(f"Operacion: {type(operacion).__name__} {operacion.a} {operacion.signo()} {operacion.b} = Error: División por cero")
    except TypeError:
        print(f"Operacion: {type(operacion).__name__} {operacion.a} {operacion.signo()} {operacion.b} = Error: Tipos de datos incompatibles")
```

```python
# Desafío 69

import math

def calcular_factorial(n):
    if not isinstance(n, int) or isinstance(n, bool):
        raise TypeError("El valor ingresado debe ser un número entero.")

    if n < 0:
        raise ValueError("El número debe ser positivo.")

    limite = 1000
    if n > limite:
        raise OverflowError(f"El número es demasiado grande. El límite máximo permitido es {limite}.")

    return math.factorial(n)


#Prueba

numeros = [5, 0, -4, 5.5, "10", 1500]

for numero in numeros:
    try:
        resultado = calcular_factorial(numero)
        print(f"El factorial de {numero} es: {resultado}")
    except TypeError as error:
        print(f"TypeError con '{numero}': {error}")
    except ValueError as error:
        print(f"ValueError con '{numero}': {error}")
    except OverflowError as error:
        print(f"OverflowError con '{numero}'': {error}")
```

```python
# Desafío 71
def leer_archivo(nombre_archivo):
    archivo = None
    try:
        archivo = open(nombre_archivo, "r", encoding="utf-8")
        contenido = archivo.read()
        print(contenido)

    except FileNotFoundError:
        print(f"Error: El archivo '{nombre_archivo}' no existe.")

    finally:
        if archivo and not archivo.closed:
            archivo.close()
            print()
            print("El archivo ha sido cerrado correctamente.")


#Prueba
leer_archivo("archivo_falso.txt")
print()
    
import os

nombre_archivo = "Texto_de_prueba.txt"
ruta = r"C:\Users\aleec\OneDrive\Escritorio\Texto_de_prueba.txt"
leer_archivo(ruta)
```
