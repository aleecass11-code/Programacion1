# Introducción a los Tipos Abstractos de Datos (TAD)

## Concepto de TAD

Un Tipo Abstracto de Datos (TAD) es una especificación de un conjunto de datos y las operaciones que se pueden realizar con esos datos. El TAD se centra en qué operaciones existen y no en cómo se implementan. Es una herramienta conceptual que nos permite abstraernos de los detalles y pensar en términos de operaciones y datos.

### Ejemplo de TAD: Lista


```python
lista = []

def agregar(elemento):
    lista.append(elemento)

def eliminar(elemento):
    if elemento in lista:
        lista.remove(elemento)

def buscar(elemento):
    return elemento in lista

#
nuevo = input(print("escribe un listado de palabras:" )).split(" ")
for palabra in nuevo:
    agregar(palabra)
print(lista) 
```

## Desventajas de este enfoque

1. **Falta de encapsulamiento**: En el ejemplo proporcionado, tanto las operaciones como los datos están expuestos y pueden ser modificados directamente desde fuera del TAD. Esto viola el principio de encapsulamiento, que es fundamental para el diseño de software robusto y mantenible.

2. **Dificultad en reutilización**: Con el diseño actual, si quisieras tener múltiples listas y aplicar operaciones sobre ellas, tendrías que crear funciones separadas o modificar las existentes, lo que no es práctico.

3. **Problemas de extensibilidad**: Cambiar la implementación interna (por ejemplo, cambiar de una lista a un árbol binario para mejorar la eficiencia) requeriría modificar todas las funciones que interactúan con la estructura de datos. Esto hace que el código sea difícil de mantener y de extender con nuevas funcionalidades.

4. **Falta de abstracción**: Aunque el TAD debería ayudarnos a abstraernos de los detalles de implementación, el enfoque actual no lo hace de manera efectiva. Esto podría llevar a errores si no se comprenden completamente las implementaciones subyacentes.

```python
#Desafío 44
biblioteca = []

def agregar(biblioteca, titulo):
    #Agrega un nuevo libro a la biblioteca.
    biblioteca.append(titulo)
    print(f'Se agregó "{titulo}" a la biblioteca.')

def eliminar(biblioteca, titulo):
    #Elimina un libro de la biblioteca.
    for libro in biblioteca:
        if libro.lower() == titulo.lower():
            biblioteca.remove(libro)
            print(f'"{libro}" se eliminó de la biblioteca.')
            return
    print(f"El libro no se puede eliminar porque no está en la biblioteca.")

def buscar(biblioteca, titulo):
    for libro in biblioteca:
        if libro.lower() == titulo.lower():
            print(f'El libro "{libro}" SÍ está disponible en la biblioteca.')
            return
    print(f'El libro "{titulo}" NO se encuentra en la biblioteca.')

def mostrar(biblioteca):
    #Muestra los libros disponibles.
    if not biblioteca:
        print("\nLa biblioteca está vacía.")
        return
    else:
        print("\n   Libros Disponibles")
    for disponibles, libro in enumerate(biblioteca, 1):
        print(f"{disponibles}. {libro}")

#Menú de prueba
print("\n    MENÚ BIBLIOTECA")
print("1) Agregar un libro")
print("2) Eliminar un libro")
print("3) Mostrar la biblioteca")
print("4) Consultar disponibilidad")
print("5) Finalizar el programa\n")

while True:
    opcion = input("Selecciona una opción (1-5): ").strip() 
    #strip() sirve para eliminar los espacios al comienzo y al final de los que se escribe

    if opcion == "1":
        titulo = input("Ingresa el título del libro a agregar: ").strip()
        if titulo:
            agregar(biblioteca, titulo)
        else:
            print("El título no puede estar vacío.")

    elif opcion == "2":
        titulo = input("Ingresa el título del libro a eliminar: ").strip()
        if len(biblioteca) == 0:
            print("La biblioteca está vacía")
        elif titulo:
            eliminar(biblioteca, titulo)
        else:
            print("El título no puede estar vacío.")

    elif opcion == "3":
        mostrar(biblioteca)

    elif opcion == "4":
        titulo = input("Ingresa el título del libro a consultar: ").strip()
        if titulo:
            buscar(biblioteca, titulo)
        else:
            print("El título no puede estar vacío.")
    
    elif opcion == "5":
        break

    else:
        print("Opción no válida. Por favor, ingresa un número del 1 al 5.")
```

```python
#Desafío 45
#Supón que estás desarrollando un juego de video con distintos tipos de personajes y armas. Los requerimientos cambian con frecuencia, añadiendo nuevos personajes y habilidades. Mantener y actualizar TADs en este escenario podría ser una tarea titánica.

def crear_personaje(nombre, tipo, habilidades):
    return {
        "personaje": nombre,
        "tipo": tipo,
        "habilidades": habilidades
    }
    
def crear_arma(nombre, tipo, daño):
    return {
        "arma": nombre,
        "tipo": tipo,
        "daño": daño
    }
    
def mostrar_personaje(personaje):
    print(f"Personaje: {personaje['personaje']}")
    print(f"Tipo: {personaje['tipo']}")
    print(f"Habilidades: {', '.join(personaje['habilidades'])}")

def mostrar_arma(arma):
    print(f"Arma: {arma['arma']}")
    print(f"Tipo: {arma['tipo']}")
    print(f"Daño: {arma['daño']}")
    
def agregar_habilidad(personaje, habilidad):
    personaje['habilidades'].append(habilidad)
    
def potenciar_arma(arma, incremento):
    arma['daño'] += incremento
    
personaje1 = crear_personaje("Thor", "Guerrero", ["Rayo"])
arma1 = crear_arma("Martillo", "Mágica", 100)


mostrar_personaje(personaje1)
mostrar_arma(arma1)

agregar_habilidad(personaje1, "Vuelo")
potenciar_arma(arma1, 50)
print("\nPersonaje actualizado:")
mostrar_personaje(personaje1)
print("\nArma actualizada:")
mostrar_arma(arma1)
```
